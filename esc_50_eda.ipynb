```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


data_path = ""



```

```python
df = pd.read_csv(data_path)
print("Data loaded:", df.shape)
df.head()
```

```python
# Columns and types
df.info()

# Class distribution
print("\nNumber of unique classes:", df["category"].nunique())
print("Number of recordings:", len(df))

# Dataset size per class
df["category"].value_counts().sort_values(ascending=False).plot(kind="bar", figsize=(12, 4), title="Class Distribution")
plt.xlabel("Category")
plt.ylabel("Number of Clips")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Fold distribution
sns.countplot(x="fold", data=df)
plt.title("Distribution of Folds")
plt.show()

# category vs fold
pd.crosstab(df["category"], df["fold"])
```

```python
# duplicated filenames
print("Duplicate filenames:", df["filename"].duplicated().sum())

# missing values
print("Missing values:\n", df.isnull().sum())
```

```python
# Category counts
class_counts = df["category"].value_counts().sort_index()
print("Samples per class:\n", class_counts)

# Stats
print("\nMin samples per class:", class_counts.min())
print("Max samples per class:", class_counts.max())
print("Mean samples per class:", class_counts.mean())
```
